```python
# Q35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# Prediction: [[1, 99], [3, 4]]
# Explanation: A shallow copy shares inner lists, so modifying an inner element through b also changes a.
```

```python
# Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

# Prediction: [[1, 2], [3, 4]]
# Explanation: A deep copy creates independent nested lists, so modifying b does not affect a.
```

```python
# Q37
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)

# Prediction: [[10], [20, 30]]
# Explanation: The shallow copy shares inner lists, so appending to a[1] also changes the corresponding list in b.
```

```python
# Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)

# Prediction: [[10], [20]]
# Explanation: Deep copying creates independent inner lists, so changes to a do not affect b.
```

```python
# Q39
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])

# Prediction: False, True
# Explanation: A shallow copy creates a new outer list but shares the original inner lists.
```

```python
# Q44
def test(*args):
    print(args)

test(1, 2)

# Prediction: (1, 2)
# Explanation: *args collects all positional arguments into a tuple.
```

```python
# Q45
def test(*args):
    print(type(args))

test("a", "b", "c")

# Prediction: <class 'tuple'>
# Explanation: The *args parameter always collects positional arguments into a tuple.
```

```python
# Q46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)

# Prediction: {'name': 'Ram', 'age': 25}
# Explanation: **kwargs collects keyword arguments into a dictionary of key-value pairs.
```
