## Q1. Find the Shortest Word in a Sentence

**1. Understand the problem:** Restate what the program needs to find

**2. Identify input and output:** Input = `Python makes coding easy`; expected output = `easy`.

**3. Break it into smaller parts:** Split the sentence into words, then compare word lengths or count how often each word occurs, depending on the question.

**4. Plan the logic:** Keep the first valid word as the current answer. Replace it only when a better candidate is found; a strict comparison preserves the first word in a tie.

**5. Implement the solution:** Put the logic into a function so it can be reused with different inputs.

**6. Test and verify:** Run the given example and the additional case `I love Python`. Compare the printed results with the expected outputs for both test cases.


```python
def find_shortest_word(sentence):
    words = sentence.split()
    shortest = words[0]
    for word in words:
        if len(word) < len(shortest):
            shortest = word
    return shortest


# Test 1: Given example
result = find_shortest_word("Python makes coding easy")
print("Output:", result)

# Test 2: Additional test case
result = find_shortest_word("I love Python")
print("Output:", result)
```

## Q6. Count Vowels in a String

**1. Understand the problem:** Restate what the program needs to find.

**2. Identify input and output:** Input = `Education`; expected output = `5`.

**3. Break it into smaller parts:** Visit each character and distinguish letters, vowels, spaces, and repeated characters as required.

**5. Implement the solution:** Put the logic into a function so it can be reused with different inputs.

**6. Test and verify:** Run the given example and the additional case `Hello World`. Compare the printed results with the expected outputs for both test cases.


```python
def count_vowels(text):
    vowels = "aeiou"
    count = 0
    for character in text.lower():
        if character in vowels:
            count += 1
    return count


# Test 1: Given example
result = count_vowels("Education")
print("Output:", result)

# Test 2: Additional test case
result = count_vowels("Hello World")
print("Output:", result)
```

## Q7. Count Consonants in a String

**1. Understand the problem:** Restate what the program needs to find.

**2. Identify input and output:** Input = `Python 123!`; expected output = `5`.

**3. Break it into smaller parts:** Visit each character and distinguish letters, vowels, spaces, and repeated characters as required.

**4. Plan the logic:** Use a counter or compare character frequencies. Skip spaces where instructed and add a repeated character only once.

**5. Implement the solution:** Put the logic into a function so it can be reused with different inputs.

**6. Test and verify:** Run the given example and the additional case `AI is fun.`. Compare the printed results with the expected outputs for both test cases.


```python
def count_consonants(text):
    vowels = "aeiou"
    count = 0
    for character in text.lower():
        if character.isalpha() and character not in vowels:
            count += 1
    return count


# Test 1: Given example
result = count_consonants("Python 123!")
print("Output:", result)

# Test 2: Additional test case
result = count_consonants("AI is fun.")
print("Output:", result)
```

```python
def least_frequent_character(text):
    least_character = ""
    least_count = len(text) + 1
    for character in text:
        if character != " ":
            if text.count(character) < least_count:
                least_count = text.count(character)
                least_character = character
    return least_character


# Test 1: Given example
result = least_frequent_character("success")
print("Output:", result)

# Test 2: Additional test case
result = least_frequent_character("aabc")
print("Output:", result)
```
